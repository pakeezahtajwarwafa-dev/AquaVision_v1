```python
import sys
sys.path.append('..')
import pandas as pd
import torch
from torch.utils.data import DataLoader
from config import load_config
from aquavision.dataset import AquaDataset, get_transforms
from aquavision.models import build_model

cfg = load_config('../config.yaml')
df_fish = pd.read_csv(cfg.paths.datasets / 'manifests' / 'fish_manifest_deduped.csv')
df_train = df_fish[df_fish['split'] == 'train']

transform = get_transforms(img_size=cfg.training.img_size, is_train=True)
dataset = AquaDataset(df_train, transform=transform)
loader = DataLoader(dataset, batch_size=cfg.training.batch_size, shuffle=True)

images, labels = next(iter(loader))
print(f'Batch Images Shape: {images.shape}')
print(f'Batch Labels Shape: {labels.shape}')

model = build_model('resnet34', num_classes=len(dataset.label_to_idx), pretrained=True)
outputs = model(images)
print(f'Forward Pass Output Shape: {outputs.shape}')
```
